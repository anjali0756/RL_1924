```python

gamma = 0.9
theta = 0.000001

V0 = 0
V1 = 0
iterations = 0

while True:
    new_V0 = -1 + 0.81 * V1
    new_V1 = 89.9 + 0.09 * V0

    delta = max(abs(new_V0 - V0), abs(new_V1 - V1))

    V0 = new_V0
    V1 = new_V1
    iterations += 1

    if delta < theta:
        break

Q_S0_Down = 0.1 * (-1) + 0.9 * (-1 + gamma * V1)
Q_S0_Right = -1
Q_S1_Right = 0.9 * 100 + 0.1 * (-1 + gamma * V0)
Q_S1_Up = -1 + gamma * V0

print("Number of Iterations:", iterations)

print("\nState Values:")
print("V(S0) =", round(V0, 2))
print("V(S1) =", round(V1, 2))

print("\nAction Values:")
print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

print("\nPolicy Consistency Check:")
print("V(S0) = Q(S0, Down):", round(V0, 2) == round(Q_S0_Down, 2))
print("V(S1) = Q(S1, Right):", round(V1, 2) == round(Q_S1_Right, 2))
```
